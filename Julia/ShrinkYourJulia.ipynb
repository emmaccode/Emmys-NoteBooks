```julia
# more readable version:
function sample(x::Int64, y::Float64)
    z = if x > y
         x / y
    else
         y / x
    end
    value = begin
        res = (z + x) * y
        res ^ 2 + 1
    end
    return(value, z, x + y)
end
```

```julia
# traditional version:
function sample(x::Int64, y::Float64)
    z = nothing
    if x > y
         z = x / y
    else
         z = y / x
    end
    res = (z + x) * y
    value = res ^ 2 + 1
    return(value, z, x + y)
end
```

## broadcasting

```julia
a = [1, 2, 3]
b = [4, 5, 6]

c = []
for i in 1:length(a)
    push!(c, a[i] + b[i])
end
```

## method passthroughs

```julia
function print_values(pairs::Pair{<:Any, <:Any} ...)
    for pair in pairs
        println("KEY: $(pair[1])")
        println("VALUE: $(pair[2])")
    end
    nothing::Nothing
end
```

```julia
print_values(dct::AbstractDict) = begin
    print_values(pairs(dct) ...)
end
```

```julia
function print_values(pairs::Pair{<:Any, <:Any} ...; skip_keys::Bool = false)
    # reverse skip_keys to avoid doing so on each loop
    skip_keys = ~(skip_keys)
    for pair in pairs
        if skip_keys
             println("KEY: $(pair[1])")
        end
        println("VALUE: $(pair[2])")
    end
    nothing::Nothing
end
```

```julia
print_values(dct::AbstractDict; keys ...) = begin
    print_values(pairs(dct) ...; keys ...)
end
```

## generic functions

```julia
print_values(data::Any; keys ...) = begin
    (d_names, d_vals) = try
        (names(d_names), eachcol(d_names))
    catch
         ([], [])
    end
    if ~(length(d_names) == length(d_vals))
        throw("name and column count not equal!")
    end
    print_values((k => v for (key, v) in zip(d_names, d_vals)) ...)
end
```

```julia
filler(a::Any) = return("" => "")

function find_correct_f(val::Any)::Function
    T = typeof(val)
    ms = methods(eachcol, Any[T])
    n = length(ms)
    if n < 1
        return(eachcol)
    end
    ms = methods(pairs, Any[T])
    n = length(ms)
    if n < 1
        return(pairs)
    end
    return(filler)
end
```

## abstraction

```julia
abstract type Shape end
struct Circle <: Shape
    radius::Float64
end
struct Rectangle <: Shape
    width::Float64
    height::Float64
end

area(shape::Circle) = π * shape.radius ^ 2
area(shape::Rectangle) = shape.width * shape.height

describe(shape::Shape) = println("Area: ", area(shape))
```

```julia
function getsum(vec::Vector{<:Number})
    sum(vec)
end

function getsum(vec::Vector{<:AbstractString})
    throw(" can't get the sum of strings")
end

function getsum(vec::Vector{Bool})
  sum(Vector{Int64}(vec))
end
```

## MORE Constructors

```julia
struct Point
 x::Float64
 y::Float64

 # Standard inner constructor
 function Point(x::Float64, y::Float64)
  new(x, y)
 end

 # Construct from another Point
 Point(p::Point) = new(p.x, p.y)

 # Construct from a tuple
 Point(t::Tuple{<:Real, <:Real}) = new(float(t[1]), float(t[2]))
end
```
